```python
import pandas as pd
import os

# Path to the folder containing CSVs
folder_path = "/content/drive/MyDrive/Fingat_Indian/extracted_data/processed_data/"

# List all CSV files
csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]

# Create an empty list to store DataFrames
dfs = []

# Loop through each file and read it
for file in csv_files:
    df = pd.read_csv(os.path.join(folder_path, file))

    # Add a 'ticker' column (stock ID)
    df["ticker"] = file.replace(".csv", "")  # Use filename as ticker

    # Append to the list
    dfs.append(df)

# Merge all into a single DataFrame
full_data = pd.concat(dfs, ignore_index=True)

# Convert 'date' to datetime
full_data['date'] = pd.to_datetime(full_data['date'])

# Set 'date' as index
full_data.set_index('date', inplace=True)

# Show sample rows
print(full_data.head())
print(full_data.info())
```

# Graph Construction

```python
import pandas as pd

# Load the saved data
file_path = "/content/drive/MyDrive/Fingat_Indian/extracted_data/full_data.csv"
full_data = pd.read_csv(file_path, index_col="date", parse_dates=True)

# Compute correlation matrix using 'close' prices for each stock
correlation_matrix = full_data.pivot(columns="ticker", values="close").corr()

print("✅ Stock correlation matrix calculated!")
print(correlation_matrix.head())
```

```python
import numpy as np

# Define threshold for strong connections (0.7 means 70% correlation)
threshold = 0.7

# Find stock pairs that are highly correlated
edge_list = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):  # Avoid duplicate pairs
        if correlation_matrix.iloc[i, j] > threshold:
            edge_list.append((correlation_matrix.columns[i], correlation_matrix.columns[j]))

# Convert to DataFrame
edge_df = pd.DataFrame(edge_list, columns=["stock_1", "stock_2"])

print("✅ Edge list created!")
print(edge_df.head())
```

```python
edge_save_path = "/content/drive/MyDrive/Fingat_Indian/extracted_data/stock_edges.csv"
edge_df.to_csv(edge_save_path, index=False)

print(f"✅ Edge list saved at: {edge_save_path}")
```

```python
import torch
import pandas as pd
from torch_geometric.data import Data

# Load the saved edge list
edge_save_path = "/content/drive/MyDrive/Fingat_Indian/extracted_data/stock_edges.csv"
edge_df = pd.read_csv(edge_save_path)

# Convert stock pairs to tensor format
edge_index = torch.tensor(edge_df.values, dtype=torch.long).T  # Transpose to (2, num_edges)

# Create a PyTorch Geometric graph object
graph = Data(edge_index=edge_index)

print("✅ Graph successfully created!")
print(graph)
```

```python
import numpy as np

# Load stock data
file_path = "/content/drive/MyDrive/Fingat_Indian/extracted_data/full_data.csv"
full_data = pd.read_csv(file_path, index_col="date", parse_dates=True)

# Select relevant features (excluding ticker)
feature_cols = ["open", "high", "low", "close", "volume", "VWAP", "MACD", "rsi"]
features = full_data.groupby("ticker")[feature_cols].mean().values  # Aggregate per stock

# Convert to tensor format
x = torch.tensor(features, dtype=torch.float)

# Add features to graph
graph.x = x

print("✅ Node features added to graph!")
print(graph)
```

```python
import matplotlib.pyplot as plt
import networkx as nx
import torch
import numpy as np
from sklearn.cluster import KMeans

# Convert edge_index to a NetworkX graph
def plot_stock_graph(edge_index, num_nodes):
    G = nx.Graph()
    edges = edge_index.numpy().T
    G.add_edges_from(edges)

    # Apply clustering for color coding
    kmeans = KMeans(n_clusters=min(5, num_nodes), random_state=42, n_init=10)

    # Get all unique nodes from the edge list
    all_nodes = set(edge_index.numpy().flatten())

    # Create node positions for all unique nodes
    node_positions = nx.spring_layout(G, seed=42)

    # Ensure node_labels covers all unique nodes
    node_labels = {i: str(i) for i in all_nodes}

    # Assign colors based on clusters (only for nodes with positions)
    cluster_labels = kmeans.fit_predict(np.array(list(node_positions.values())))
    colors = [plt.cm.jet(i / max(cluster_labels)) for i in cluster_labels]

    # Draw the graph using only the nodes with positions
    nodes_to_draw = list(node_positions.keys())

    plt.figure(figsize=(10, 6))
    nx.draw(G, pos=node_positions, with_labels=True, labels={node: node_labels[node] for node in nodes_to_draw},
            node_color=colors, edge_color='gray', node_size=500, font_size=10,
            nodelist=nodes_to_draw)  # Only draw specified nodes

    plt.title("Refined Stock Correlation Graph")
    plt.show()

# Example Usage (assuming edge_index and num_nodes are available)
plot_stock_graph(edge_index, num_nodes=graph.num_nodes)  # Use actual number of nodes in the graph
```

# Model defining and Training

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GATConv
from torch.nn import Linear

class FinGAT(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels, heads=2, dropout=0.2):
        super(FinGAT, self).__init__()

        # First Graph Attention Layer
        self.gat1 = GATConv(in_channels, hidden_channels, heads=heads, dropout=dropout)

        # Second Graph Attention Layer (aggregates multi-head outputs)
        self.gat2 = GATConv(hidden_channels * heads, out_channels, heads=1, concat=False, dropout=dropout)

        # Fully Connected Layer for Ranking Stocks
        self.fc = Linear(out_channels, 1)

    def forward(self, x, edge_index):
        x = F.relu(self.gat1(x, edge_index))  # GAT Layer 1
        x = self.gat2(x, edge_index)  # GAT Layer 2
        x = self.fc(x)  # Ranking Output
        return x.squeeze()  # Flatten output for ranking
```

```python
import torch

# Ensure 'y' is correctly shaped and assigned
stock_labels = full_data.groupby("ticker")["close"].mean().values
graph.y = torch.tensor(stock_labels, dtype=torch.float).view(-1, 1)  # Reshape to match PyG format

print("✅ Labels (y) assigned:", graph.y.shape)
```

```python
from torch_geometric.transforms import RandomNodeSplit

# Split into train/val/test (fixes missing labels issue)
splitter = RandomNodeSplit(num_val=0.1, num_test=0.1)
train_data = splitter(graph)
val_data = splitter(graph)
test_data = splitter(graph)

print("✅ Graph data split successfully!")
```

```python
from torch_geometric.loader import DataLoader

# Define batch size
batch_size = 32

# Create DataLoaders
train_loader = DataLoader([train_data], batch_size=batch_size, shuffle=True)
val_loader = DataLoader([val_data], batch_size=batch_size, shuffle=False)
test_loader = DataLoader([test_data], batch_size=batch_size, shuffle=False)

print(f"✅ DataLoaders created!")
```

```python
def train_fingat(model, train_loader, val_loader, epochs=250, lr=0.001):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = torch.nn.MSELoss()  # Mean Squared Error for ranking

    for epoch in range(epochs):
        model.train()
        total_loss = 0

        for data in train_loader:
            optimizer.zero_grad()
            out = model(data.x, data.edge_index)
            loss = criterion(out, data.y)  # Now y should exist
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        # Validation
        model.eval()
        val_loss = 0
        with torch.no_grad():
            for data in val_loader:
                out = model(data.x, data.edge_index)
                loss = criterion(out, data.y)
                val_loss += loss.item()

        print(f"Epoch {epoch+1}/{epochs}, Train Loss: {total_loss:.4f}, Val Loss: {val_loss:.4f}")

    print("✅ Training Complete!")
```

```python
# Initialize the model
model = FinGAT(in_channels=8, hidden_channels=16, out_channels=8)

# Train the model
train_fingat(model, train_loader, val_loader)
```

# Evaluation

```python
def evaluate_fingat(model, test_loader):
    model.eval()
    criterion = torch.nn.MSELoss()
    test_loss = 0

    with torch.no_grad():
        for data in test_loader:
            out = model(data.x, data.edge_index)
            loss = criterion(out, data.y)
            test_loss += loss.item()

    print(f"✅ Test Loss: {test_loss:.4f}")

# Run evaluation
evaluate_fingat(model, test_loader)
```

Now, let's create a basic Flask API application. This script will be runnable outside of this Colab notebook, and you can upload it to your GitHub repository. It will load the trained model and the ticker mapping, then expose a `/predict` endpoint.

```python
%%writefile app.py

import torch
import pandas as pd
import json
from flask import Flask, request, jsonify
from torch_geometric.nn import GATConv
from torch.nn import Linear
import torch.nn.functional as F

# Define the FinGAT model (must be the same as in training)
class FinGAT(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels, heads=2, dropout=0.2):
        super(FinGAT, self).__init__()
        self.gat1 = GATConv(in_channels, hidden_channels, heads=heads, dropout=dropout)
        self.gat2 = GATConv(hidden_channels * heads, out_channels, heads=1, concat=False, dropout=dropout)
        self.fc = Linear(out_channels, 1)

    def forward(self, x, edge_index):
        x = F.relu(self.gat1(x, edge_index))
        x = self.gat2(x, edge_index)
        x = self.fc(x)
        return x.squeeze()

app = Flask(__name__)

# --- Configuration --- #
MODEL_PATH = "/content/drive/MyDrive/Fingat_Indian/extracted_data/fingat_model.pth"
TICKER_MAP_PATH = "/content/drive/MyDrive/Fingat_Indian/extracted_data/ticker_to_id.json"

# --- Load Model and Ticker Map --- #
def load_resources():
    global model, ticker_to_id_map, id_to_ticker_map

    # Load model
    model = FinGAT(in_channels=8, hidden_channels=16, out_channels=8) # Ensure params match training
    model.load_state_dict(torch.load(MODEL_PATH, map_location=torch.device('cpu')))
    model.eval() # Set model to evaluation mode

    # Load ticker mapping
    with open(TICKER_MAP_PATH, 'r') as f:
        ticker_to_id_map = json.load(f)
    id_to_ticker_map = {v: k for k, v in ticker_to_id_map.items()}

    print("Model and ticker mappings loaded successfully.")

# Load resources when the app starts
with app.app_context():
    load_resources()


@app.route('/')
def home():
    return "FinGAT Stock Prediction API is running!"

@app.route('/predict', methods=['POST'])
def predict():
    try:
        # Here, we assume the API consumer provides features for ALL stocks
        # in the order of ticker IDs and the edge_index for the graph structure.
        # In a real-world scenario, you might have a more complex input or
        # pre-process data within the API based on requested tickers.
        data = request.get_json(force=True)
        features_list = data['features'] # List of lists, e.g., [[f1_s1, ..., f8_s1], [f1_s2, ..., f8_s2], ...]
        edge_index_list = data['edge_index'] # List of [source, target] pairs

        # Convert to PyTorch tensors
        features = torch.tensor(features_list, dtype=torch.float)
        edge_index = torch.tensor(edge_index_list, dtype=torch.long).T # Transpose for PyG

        # Create a dummy graph for prediction (only x and edge_index are needed here)
        # Note: In a production setting, you'd likely rebuild the graph with current data
        # or adjust inputs to match the model's expectation.
        from torch_geometric.data import Data
        graph_input = Data(x=features, edge_index=edge_index)

        with torch.no_grad():
            predictions = model(graph_input.x, graph_input.edge_index).numpy()

        # Map predictions back to original ticker names
        ranked_stocks = []
        for i, score in enumerate(predictions):
            ticker_name = id_to_ticker_map.get(i, f"Unknown_{i}") # Handle potential missing keys
            ranked_stocks.append({'ticker': ticker_name, 'score': float(score)})

        # Sort by score for ranking
        ranked_stocks.sort(key=lambda x: x['score'], reverse=True)

        return jsonify(ranked_stocks)

    except Exception as e:
        return jsonify({'error': str(e)}), 400

if __name__ == '__main__':
    # This is for running locally for development/testing
    # For deployment, a production WSGI server like Gunicorn is recommended.
    app.run(host='0.0.0.0', port=5000, debug=True)
```

```python
def predict_and_rank_stocks(model, graph):
    model.eval()

    with torch.no_grad():
        stock_scores = model(graph.x, graph.edge_index)  # Get scores for each stock

    # Convert scores to a pandas DataFrame
    stock_ranking = pd.DataFrame({
        "ticker": list(ticker_to_id.keys()),  # Convert IDs back to tickers
        "score": stock_scores.cpu().numpy()  # Convert tensor to numpy
    })

    # Sort by predicted score (higher = better)
    stock_ranking = stock_ranking.sort_values(by="score", ascending=False)

    print("✅ Stock Predictions & Rankings:")
    print(stock_ranking)

# Run predictions
predict_and_rank_stocks(model, graph)
```
